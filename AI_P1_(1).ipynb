Import Libraries

```python
import kagglehub
import pandas as pd
import os

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn import svm

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

Download the Dataset

```python
path = kagglehub.dataset_download(
    "uciml/sms-spam-collection-dataset"
)

print("Path to dataset files:", path)
```

Check the Dataset Files

```python
print(os.listdir(path))
```

Load the Dataset

```python
file_path = os.path.join(path, "spam.csv")

df = pd.read_csv(
    file_path,
    encoding="latin-1"
)

df.head()
```

Keep Only the Required Columns

```python
df = df[["v1", "v2"]]

df.columns = ["label", "message"]

df.head()
```

Check Dataset Information

```python
print("Dataset Shape:", df.shape)

print("\nLabel Count:")
print(df["label"].value_counts())

print("\nMissing Values:")
print(df.isnull().sum())
```

Convert Labels into Numbers

```python
df["label"] = df["label"].map({
    "ham": 0,
    "spam": 1
})

df.head()
```

Define Input and Target

```python
X = df["message"]
y = df["label"]

print("Input Data:")
print(X.head())

print("\nTarget:")
print(y.head())
```

Split the Dataset

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Total Data:", len(X))
print("Training Data:", len(X_train))
print("Testing Data:", len(X_test))
```

View Training Data

```python
print("Training Messages:")
print(X_train.head())

print("\nTraining Labels:")
print(y_train.head())
```

TF-IDF Vectorization

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    max_features=5000
)

X_train_vector = vectorizer.fit_transform(X_train)

X_test_vector = vectorizer.transform(X_test)
```

Check Vectorized Data

```python
print("Training Vector Shape:")
print(X_train_vector.shape)

print("\nTesting Vector Shape:")
print(X_test_vector.shape)

print("\nOriginal Message:")
print(X_train.iloc[0])

print("\nVectorized Message:")
print(X_train_vector[0])
```

Create the SVM Model

```python
model = svm.SVC(
    kernel="linear"
)

model
```

Train the SVM Model

```python
model.fit(
    X_train_vector,
    y_train
)

print("Model training completed.")
```

Predict the Test Data

```python
y_pred = model.predict(
    X_test_vector
)

print(y_pred[:20])
```

Compare Actual and Predicted Values

```python
results = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

results.head(20)
```

Calculate Accuracy

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:", accuracy)

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)
```

Classification Report

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Ham", "Spam"]
    )
)
```

Confusion Matrix

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

print("Confusion Matrix:")
print(cm)
```

Testing with Own SMS

```python
new_message = [
    "Congratulations! You have won a free prize. Call now!"
]

new_vector = vectorizer.transform(
    new_message
)

prediction = model.predict(
    new_vector
)

print(prediction)
```

Display Spam or Ham

```python
if prediction[0] == 1:
    print("SPAM MESSAGE")
else:
    print("HAM MESSAGE")
```

Test Another Normal Message

```python
new_message = [
    "Hi, are you coming to college today?"
]

new_vector = vectorizer.transform(
    new_message
)

prediction = model.predict(
    new_vector
)

if prediction[0] == 1:
    print("SPAM")
else:
    print("HAM")
```
